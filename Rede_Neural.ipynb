```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tensorflow import keras

# 1. Carregar os dados (treino e teste)
(x_treino, y_treino), (x_teste, y_teste) = keras.datasets.mnist.load_data()

# 2. Normalizar os pixels para ficarem entre 0 e 1
x_treino = x_treino / 255.0
x_teste = x_teste / 255.0

# 3. Criar o modelo básico
modelo = keras.Sequential(
    [
        keras.layers.Flatten(
            input_shape=(28, 28)
        ),  # Transforma a imagem em linha
        keras.layers.Dense(
            128, activation="relu"
        ),  # Camada de aprendizado
        keras.layers.Dense(10, activation="softmax"),  # 10 opções (0 a 9)
    ]
)

modelo.compile(
    optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"]
)

# 4. Treinar o modelo
modelo.fit(x_treino, y_treino, epochs=5)

# 5. Testar com a primeira imagem do teste
i = 0
predicoes = modelo.predict(x_teste[i : i + 1])
previsto = np.argmax(predicoes[0])

print(f"O modelo previu que é o número: {previsto}")
print(f"O número real da imagem é: {y_teste[i]}")

# Mostrar a imagem
plt.imshow(x_teste[i], cmap="gray")
plt.title(f"Previsto: {previsto} | Real: {y_teste[i]}")
plt.axis("off")
plt.show()
```
